```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração do estilo visual do gráfico
plt.figure(figsize=(10, 6))
sns.set_theme(style="whitegrid")

# Criando o Boxplot para analisar a distribuição do Salário por Departamento
sns.boxplot(data=df_rh, x='Departamento', y='Salario', palette='Set2')

# Títulos e rótulos
plt.title('Distribuição de Salários por Departamento', fontsize=14, fontweight='bold')
plt.xlabel('Departamento', fontsize=12)
plt.ylabel('Salário (R$)', fontsize=12)
plt.xticks(rotation=45)

# Ajustando o layout
plt.tight_layout()

# Salvando o gráfico como uma imagem PNG
plt.savefig('boxplot_salarios_departamento.png', dpi=300)

# Exibindo o gráfico
plt.show()
```

```python
# Calculando a mediana salarial por departamento
tabela_medianas = df_rh.groupby('Departamento')['Salario'].median().reset_index()

# Renomeando as colunas para ficar com uma apresentação limpa
tabela_medianas.columns = ['Departamento', 'Mediana Salarial (R$)']

# Formatando os valores como moeda para melhor visualização
tabela_medianas['Mediana Salarial (R$)'] = tabela_medianas['Mediana Salarial (R$)'].map('R$ {:,.2f}'.format)

# Exibindo a tabela gerada
display(tabela_medianas)

# Salvando a tabela de medianas em um arquivo CSV
tabela_medianas.to_csv('medianas_salariais.csv', index=False, encoding='utf-8')

print("Tabela de medianas salva com sucesso como 'ex1_medianas_salariais.csv'!")
```
